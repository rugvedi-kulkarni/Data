```python
import json
```

```python
# Load json data
def load_data(filename):
    with open("store_data.json", "r") as f:
        data = json.load(f)

    return data
```

```python
# call the function
data = load_data("store_data.json")
print(type(data), data)
```

```python
# Clean and Structure them

def clean_data(data):
    
    text_to_num = {"one": 1, "two": 2, "three": 3, "four": 4, "five": 5} # dictionary
    cleaned_data = [] # empty list
    unique_users = set() #empty set
    
    for user in data: # data => list(sequence)
        
        # Clean rating data - data consistency
        raw_rating = str(user["rating"]).strip().lower() # user => dictionary, .strip(), .lower() => string methods
       
        if raw_rating in text_to_num: # ex, "four" in text_to_num: raw_rating = text_to_num["four"] => 4
            raw_rating = text_to_num[raw_rating]
            
       
        user["rating"] = raw_rating

        # handle missing values
        raw_age = user.get("age") #.get() => dict method
        if(raw_age == None):
            raw_age = None

        # Deduplication
        if user["name"].strip() in unique_users: # user => dict, dict["key"], Alice => string, .strip() => string method
            continue 
            # initially the set was empty first it will check whether the name exists in set or not if its not then add the name in set

        else:
            unique_users.add(user["name"]) # .add() => set method
            cleaned_data.append(user) # add cleaned data to the list
            
    return cleaned_data
        
```

```python
res = clean_data(data)
```

```python
# Get Meaningful insights

def get_insights(data):

    tot_rating = 0

    for user in data:
        tot_rating += float(user["rating"])

    print(f'Average rating: {tot_rating/len(data)}')

    # percentage of users with poor rating
    poor_rating = 0

    for user in data:
        if float(user["rating"]) < 3:
            poor_rating += 1

    print(f'% of Poor rating is {poor_rating/len(data) * 100}')
```

```python
get_insights(res)
```
